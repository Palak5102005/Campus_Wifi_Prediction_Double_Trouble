```python
import pandas as pd

survey = pd.read_csv("/kaggle/input/datasets/psrc0514/responses/Responses - Form responses 1.csv")

print(survey.shape)
display(survey.head())
```

```python
survey.columns = [
    "timestamp",
    "frequency",
    "problem_type",
    "location",
    "impact",
    "desired_solution"
]

survey["location"] = (
    survey["location"]
    .astype(str)
    .str.strip()
    .str.lower()
)
```

```python
def normalize_location(x):

    if "hostel room" in x:
        return "HOSTEL ROOM"

    if "hostel common" in x:
        return "HOSTEL COMMON AREA"

    if "academic" in x or "classroom" in x:
        return "ACADEMIC BUILDINGS"

    if "library" in x:
        return "LIBRARY"

    if "mess" in x or "canteen" in x:
        return "MESS / CANTEEN"

    if "new sac" in x:
        return "NEW SAC"

    if "everywhere" in x or "every above" in x or "all places" in x:
        return "EVERYWHERE"

    return "OTHER"

survey["location_group"] = survey["location"].apply(
    normalize_location
)
```

```python
location_survey = (
    survey
    .groupby("location_group")
    .size()
    .reset_index(name="responses")
    .sort_values("responses", ascending=False)
)

display(location_survey)
```

```python
problem_survey = (
    survey
    .groupby("problem_type")
    .size()
    .reset_index(name="responses")
    .sort_values("responses", ascending=False)
)

display(problem_survey)
```

```python
solution_demand = (
    survey
    .groupby("desired_solution")
    .size()
    .reset_index(name="responses")
    .sort_values("responses", ascending=False)
)

display(solution_demand)
```

```python
def expand_location(x):
    x = str(x).strip().lower()

    if x in ["everywhere", "every above option", "all places"]:
        return [
            "HOSTEL ROOM",
            "HOSTEL COMMON AREA",
            "ACADEMIC BUILDINGS",
            "LIBRARY",
            "MESS / CANTEEN"
        ]

    if "hostel room" in x:
        return ["HOSTEL ROOM"]

    if "hostel common" in x:
        return ["HOSTEL COMMON AREA"]

    if "academic" in x or "classroom" in x:
        return ["ACADEMIC BUILDINGS"]

    if "library" in x:
        return ["LIBRARY"]

    if "mess" in x or "canteen" in x:
        return ["MESS / CANTEEN"]

    return ["OTHER"]
```

```python
survey_location = survey.copy()

survey_location["location_group"] = survey_location["location"].apply(
    expand_location
)

survey_location = survey_location.explode("location_group")

display(
    survey_location[
        ["location", "location_group", "frequency", "problem_type", "impact"]
    ].head(20)
)
```

```python
location_profile = (
    survey_location
    .groupby("location_group")
    .size()
    .reset_index(name="survey_mentions")
    .sort_values("survey_mentions", ascending=False)
)

display(location_profile)
```

```python
frequency_score = {
    "Almost every day": 4,
    "Several times a week": 3,
    "Occasionally": 2,
    "Rarely": 1
}

survey_location["frequency_score"] = (
    survey_location["frequency"]
    .map(frequency_score)
)
```

```python
frequency_profile = (
    survey_location
    .groupby("location_group")
    .agg(
        avg_frequency_score=("frequency_score", "mean")
    )
    .reset_index()
)

display(frequency_profile)
```

```python
def expand_problem(x):
    x = str(x).strip().lower()

    if "all" in x or "every" in x:
        return [
            "SLOW SPEED",
            "UNABLE TO CONNECT",
            "FREQUENT DISCONNECTION"
        ]

    if "slow" in x:
        return ["SLOW SPEED"]

    if "unable" in x or "connect" in x:
        return ["UNABLE TO CONNECT"]

    if "disconnect" in x:
        return ["FREQUENT DISCONNECTION"]

    return ["OTHER"]
```

```python
survey_problem = survey_location.copy()

survey_problem["problem_group"] = survey_problem[
    "problem_type"
].apply(expand_problem)

survey_problem = survey_problem.explode("problem_group")
```

```python
problem_profile = (
    survey_problem
    .groupby(["location_group", "problem_group"])
    .size()
    .reset_index(name="mentions")
)

display(
    problem_profile.sort_values(
        ["location_group", "mentions"],
        ascending=[True, False]
    )
)
```
